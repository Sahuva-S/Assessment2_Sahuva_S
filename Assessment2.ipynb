```python
df["Average_Technical_Score"] = df[
    ["Python_Score", "SQL_Score", "Pandas_Score", "NumPy_Score"]
].mean(axis=1)

df["Performance"] = np.select(
    [
        df["Average_Technical_Score"] >= 90,
        df["Average_Technical_Score"] >= 75,
        df["Average_Technical_Score"] >= 60
    ],
    [
        "Excellent",
        "Good",
        "Average"
    ],
    default="Needs Improvement"
)
score_columns = [
    "Python_Score",
    "SQL_Score",
    "Pandas_Score",
    "NumPy_Score"
]

for col in score_columns:
    df[col] = df[col].fillna(df[col].mean())

print(df)
```

Q17

```python
#17.1
sales = pd.read_csv("sales.csv")
print(sales.head())
```

```python
#17.2
sales["Order_Date"] = pd.to_datetime(
    sales["Order_Date"],
    format="mixed"
)
```

```python
for col in ["Product", "Category", "Region"]:
    sales[col] = sales[col].str.strip()
```

```python
for col in ["Sales", "Quantity", "Discount", "Profit"]:
    sales[col] = (
        sales[col]
        .astype(str)
        .str.replace(",", "", regex=False)
        .str.replace("%", "", regex=False)
    )

    sales[col] = pd.to_numeric(
        sales[col],
        errors="coerce"
    )
```

```python
for col in ["Sales", "Quantity", "Discount", "Profit"]:
    sales[col] = sales[col].fillna(sales[col].mean())
for col in ["Region", "Product", "Category"]:
    sales[col] = sales[col].fillna(sales[col].mode()[0])
```

Q17B

```python
#q17.b.1
total_sales = sales["Sales"].sum()
print("Total Sales:", total_sales)
```

```python
#q17.b.2
total_profit = sales["Profit"].sum()
print("Total Profit:", total_profit)
```

```python
#q17.b.3
average_sales = sales["Sales"].mean()
print("Average Sales:", average_sales)
```

```python
#q17.b.4
top_products = (
    sales.groupby("Product")["Sales"].sum()
    .sort_values(ascending=False)
    .head(5)
)

print(top_products)
```

```python
#q17.5
highest_sales = (
    sales.groupby("Region")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

print(highest_sales)
```

Q17 C

```python
category_summary = sales.groupby("Category").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Average_Discount=("Discount", "mean")
)

category_summary = category_summary.sort_values(
    "Total_Sales",
    ascending=False
)

print(category_summary)
```
